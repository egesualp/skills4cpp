```python
import json
import pandas as pd

# Load the JSON file and convert to DataFrame
with open('/dss/dsshome1/02/ra95kix2/thesis/skills4cpp/results/decorte_jobbert_v2_baseline/job_title_skills_master.json', 'r') as f:
    data = json.load(f)

# Convert to DataFrame
df_master_skills = pd.DataFrame(data)
```

```python
# Expand the skills column into multiple rows
df_master_skills_expanded = df_master_skills.explode('skills').reset_index()

# Extract skill information into separate columns
df_master_skills_expanded['skill'] = df_master_skills_expanded['skills'].apply(lambda x: x['skill'] if isinstance(x, dict) else None)
df_master_skills_expanded['score'] = df_master_skills_expanded['skills'].apply(lambda x: x['score'] if isinstance(x, dict) else None)
df_master_skills_expanded['skillUri'] = df_master_skills_expanded['skills'].apply(lambda x: x['skillUri'] if isinstance(x, dict) else None)

# Drop the original skills column
df_master_skills_expanded = df_master_skills_expanded.drop('skills', axis=1)

# Rename the index column to track original rows
df_master_skills_expanded = df_master_skills_expanded.rename(columns={'index': 'original_row_index'})

df_master_skills_expanded.head()
```

```python
df_master_skills_expanded.groupby('job_title').original_row_index.nunique().sort_values()
```

```python
df_master_skills_expanded.query("job_title == 'Intern' and skill == 'conduct field work'")
```

```python
df_master_skills_expanded = df_master_skills_expanded.drop_duplicates(subset=['job_title', 'skillUri'])
```

```python
df_master_skills_expanded.to_csv('/dss/dsshome1/02/ra95kix2/thesis/skills4cpp/results/decorte_jobbert_v2_baseline/job_title_skills_master.csv', index=False)
```
