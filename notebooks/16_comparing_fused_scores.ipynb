```python
import json

# Load the JSON files
with open('/dss/dssmcmlfs01/pr74ze/pr74ze-dss-0001/ra95kix2/outputs/v5_fused_scorer/multiplicative_3/fused_predictions.json', 'r') as f:
    data_jobbert = json.load(f)

with open('/dss/dssmcmlfs01/pr74ze/pr74ze-dss-0001/ra95kix2/outputs/v5_fused_scorer/multiplicative_h2_pjmath/fused_predictions.json', 'r') as f:
    data_pjmath = json.load(f)

# Extract unique skills
unique_skills_jobbert = set()
unique_skills_pjmath = set()

# Collect skill counts and confidence scores per job
skill_counts_jobbert = []
skill_counts_pjmath = []
avg_confidences_jobbert = []
avg_confidences_pjmath = []

# Process JobBERT data - each key has a list of [skill_uri, confidence_score]
for job_id, skills_list in data_jobbert.items():
    skill_counts_jobbert.append(len(skills_list))
    confidences = [skill[1] for skill in skills_list]  # skill[1] is the confidence score
    if confidences:
        avg_confidences_jobbert.append(np.mean(confidences))
    for skill in skills_list:
        unique_skills_jobbert.add(skill[0])  # skill[0] is the skill URI

# Process PJMath data - each key has a list of [skill_uri, confidence_score]
for job_id, skills_list in data_pjmath.items():
    skill_counts_pjmath.append(len(skills_list))
    confidences = [skill[1] for skill in skills_list]  # skill[1] is the confidence score
    if confidences:
        avg_confidences_pjmath.append(np.mean(confidences))
    for skill in skills_list:
        unique_skills_pjmath.add(skill[0])  # skill[0] is the skill URI

# Print comparison statistics
print("=" * 80)
print("COMPARISON OF FUSED PREDICTIONS")
print("=" * 80)

print("\n--- UNIQUE SKILLS ---")
print(f"JobBERT (multiplicative_3): {len(unique_skills_jobbert)} unique skills")
print(f"PJMath (multiplicative_h2_pjmath): {len(unique_skills_pjmath)} unique skills")
print(f"Skills in both: {len(unique_skills_jobbert & unique_skills_pjmath)}")
print(f"Skills only in JobBERT: {len(unique_skills_jobbert - unique_skills_pjmath)}")
print(f"Skills only in PJMath: {len(unique_skills_pjmath - unique_skills_jobbert)}")

print("\n--- SKILL COUNT PER JOB ---")
print(f"JobBERT:")
print(f"  Mean: {np.mean(skill_counts_jobbert):.2f}")
print(f"  Median: {np.median(skill_counts_jobbert):.2f}")
print(f"  25th percentile: {np.percentile(skill_counts_jobbert, 25):.2f}")
print(f"  75th percentile: {np.percentile(skill_counts_jobbert, 75):.2f}")
print(f"  Min: {np.min(skill_counts_jobbert)}")
print(f"  Max: {np.max(skill_counts_jobbert)}")

print(f"\nPJMath:")
print(f"  Mean: {np.mean(skill_counts_pjmath):.2f}")
print(f"  Median: {np.median(skill_counts_pjmath):.2f}")
print(f"  25th percentile: {np.percentile(skill_counts_pjmath, 25):.2f}")
print(f"  75th percentile: {np.percentile(skill_counts_pjmath, 75):.2f}")
print(f"  Min: {np.min(skill_counts_pjmath)}")
print(f"  Max: {np.max(skill_counts_pjmath)}")

print("\n--- AVERAGE CONFIDENCE SCORE PER JOB ---")
print(f"JobBERT:")
print(f"  Mean: {np.mean(avg_confidences_jobbert):.4f}")
print(f"  Median: {np.median(avg_confidences_jobbert):.4f}")
print(f"  25th percentile: {np.percentile(avg_confidences_jobbert, 25):.4f}")
print(f"  75th percentile: {np.percentile(avg_confidences_jobbert, 75):.4f}")
print(f"  Min: {np.min(avg_confidences_jobbert):.4f}")
print(f"  Max: {np.max(avg_confidences_jobbert):.4f}")

print(f"\nPJMath:")
print(f"  Mean: {np.mean(avg_confidences_pjmath):.4f}")
print(f"  Median: {np.median(avg_confidences_pjmath):.4f}")
print(f"  25th percentile: {np.percentile(avg_confidences_pjmath, 25):.4f}")
print(f"  75th percentile: {np.percentile(avg_confidences_pjmath, 75):.4f}")
print(f"  Min: {np.min(avg_confidences_pjmath):.4f}")
print(f"  Max: {np.max(avg_confidences_pjmath):.4f}")

print("\n" + "=" * 80)
```

```python
def calculate_ranking_metrics(scores, true_indices, k_values=[1, 5, 10, 20]):
    """
    Calculate MRR and Recall@K metrics.
    """
    n_samples = len(true_indices)
    rr_sum = 0
    hits_at_k = {k: 0 for k in k_values}
    valid_samples = 0
    
    for i, true_idx in enumerate(true_indices):
        if true_idx < 0: # Skip unmapped targets
            continue
            
        valid_samples += 1
        sample_scores = scores[i]
        
        # Sort indices in descending order of score
        sorted_indices = np.argsort(sample_scores)[::-1]
        
        # Calculate Rank (1-indexed)
        rank = np.where(sorted_indices == true_idx)[0][0] + 1
        rr_sum += 1.0 / rank
        
        # Calculate Hits@K
        for k in k_values:
            if true_idx in sorted_indices[:k]:
                hits_at_k[k] += 1
                
    metrics = {
        'MRR': rr_sum / valid_samples if valid_samples > 0 else 0.0
    }
    for k in k_values:
        metrics[f'R@{k}'] = hits_at_k[k] / valid_samples if valid_samples > 0 else 0.0
        
    return metrics

def evaluate(skill_overlap_data):
# Calculate metrics if the required keys exist
    if 'scores' in skill_overlap_data and 'true_target_indices' in skill_overlap_data:
        scores = skill_overlap_data['scores']
        true_indices = skill_overlap_data['true_target_indices']
        
        metrics = calculate_ranking_metrics(scores, true_indices)
        
        print("\nRanking Metrics:")
        for metric, value in metrics.items():
            print(f"{metric}: {value:.4f}")
    else:
        print("Required keys ('scores' and 'true_target_indices') not found in data.")
        print(f"Available keys: {list(skill_overlap_data.keys())}")
    
    # Calculate metrics if the required keys exist
    if 'scores' in skill_overlap_data and 'true_target_indices' in skill_overlap_data:
        scores = skill_overlap_data['scores']
        true_indices = skill_overlap_data['true_target_indices']
        
        metrics = calculate_ranking_metrics(scores, true_indices)
        
        print("\nRanking Metrics:")
        for metric, value in metrics.items():
            print(f"{metric}: {value:.4f}")
    else:
        print("Required keys ('scores' and 'true_target_indices') not found in data.")
        print(f"Available keys: {list(skill_overlap_data.keys())}")
```

```python
def get_rank_with_ties(scores, true_idx):
    """
    Beraberlikte ortalama rank ver.
    scores: tüm occupation skorları
    true_idx: doğru hedefin indexi
    """
    # 'average' = beraberlere ortalama rank verir
    # rankdata küçükten büyüğe sıralar, biz büyükten küçüğe istiyoruz
    from scipy.stats import rankdata
    ranks = rankdata(-scores, method='average')
    return ranks[true_idx]

def calculate_ranking_metrics_2(scores, true_indices, k_values=[1, 5, 10, 20]):
    """
    Calculate MRR and Recall@K metrics.
    """
    n_samples = len(true_indices)
    rr_sum = 0
    hits_at_k = {k: 0 for k in k_values}
    valid_samples = 0
    
    for i, true_idx in enumerate(true_indices):
        if true_idx < 0: # Skip unmapped targets
            continue
            
        valid_samples += 1
        sample_scores = scores[i]
        
        # Sort indices in descending order of score
        sorted_indices = np.argsort(sample_scores)[::-1]
        
        # Calculate Rank (1-indexed) with tie handling
        rank = get_rank_with_ties(sample_scores, true_idx)
        rr_sum += 1.0 / rank
        
        # Calculate Hits@K
        for k in k_values:
            if true_idx in sorted_indices[:k]:
                hits_at_k[k] += 1
                
    metrics = {
        'MRR': rr_sum / valid_samples if valid_samples > 0 else 0.0
    }
    for k in k_values:
        metrics[f'R@{k}'] = hits_at_k[k] / valid_samples if valid_samples > 0 else 0.0
        
    return metrics

def evaluate_2(skill_overlap_data):
# Calculate metrics if the required keys exist
    if 'scores' in skill_overlap_data and 'true_target_indices' in skill_overlap_data:
        scores = skill_overlap_data['scores']
        true_indices = skill_overlap_data['true_target_indices']
        
        metrics = calculate_ranking_metrics_2(scores, true_indices)
        
        print("\nRanking Metrics:")
        for metric, value in metrics.items():
            print(f"{metric}: {value:.4f}")
    else:
        print("Required keys ('scores' and 'true_target_indices') not found in data.")
        print(f"Available keys: {list(skill_overlap_data.keys())}")
    
    # Calculate metrics if the required keys exist
    if 'scores' in skill_overlap_data and 'true_target_indices' in skill_overlap_data:
        scores = skill_overlap_data['scores']
        true_indices = skill_overlap_data['true_target_indices']
        
        metrics = calculate_ranking_metrics_2(scores, true_indices)
        
        print("\nRanking Metrics:")
        for metric, value in metrics.items():
            print(f"{metric}: {value:.4f}")
    else:
        print("Required keys ('scores' and 'true_target_indices') not found in data.")
        print(f"Available keys: {list(skill_overlap_data.keys())}")
```

```python
# Check score distribution for TRUE targets vs random targets

# For each sample, compare:
# - Score assigned to true target occupation
# - Mean score assigned to all other occupations

def analyze_target_vs_others(scores_dict):
    scores = scores_dict['scores']  # (n_samples, n_targets)
    true_indices = scores_dict['true_target_indices']
    
    target_scores = []
    other_scores = []
    
    for i, true_idx in enumerate(true_indices):
        if true_idx < 0:
            continue
        target_scores.append(scores[i, true_idx])
        # Mean of non-target scores
        mask = np.ones(scores.shape[1], dtype=bool)
        mask[true_idx] = False
        other_scores.append(scores[i, mask].mean())
    
    target_scores = np.array(target_scores)
    other_scores = np.array(other_scores)
    
    print(f"  True target score - mean: {target_scores.mean():.4f}")
    print(f"  Other targets score - mean: {other_scores.mean():.4f}")
    print(f"  Discrimination gap: {(target_scores - other_scores).mean():.4f}")
    
print("JobBERT:")
analyze_target_vs_others(jb_scores)
print("\nPJMath:")
analyze_target_vs_others(pj_scores)
```
