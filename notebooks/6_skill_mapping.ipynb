```python
import pandas as pd

# Read existing results
df = pd.read_csv('../results/cpp/experiment_results.csv')

# Add optimizer column if it doesn't exist (all past runs used Adam)
if 'optimizer' not in df.columns:
    df['optimizer'] = 'adam'
    
# Add optimizer_details if needed
if 'optimizer_details' not in df.columns:
    # For old runs, construct from existing lr (weight_decay was 0 by default)
    df['optimizer_details'] = df.apply(
        lambda row: f"Adam(lr={row['lr']:.6f}, weight_decay=0.000000)", 
        axis=1
    )

# Save back
df.to_csv('../results/cpp/experiment_results.csv', index=False)
```
