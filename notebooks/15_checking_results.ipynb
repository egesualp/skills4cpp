```python
import pickle

text_path = '/dss/dssmcmlfs01/pr74ze/pr74ze-dss-0001/ra95kix2/results/cpp/decorte/job_title_desc/scores/test_scores_text.pkl'
hybrid_path = '/dss/dssmcmlfs01/pr74ze/pr74ze-dss-0001/ra95kix2/results/cpp/decorte_static/job_titles_skills_pjmath_desc_weighted_idf_logpool_adv_2/scores/test_scores_text.pkl'
skill_path = '/dss/dssmcmlfs01/pr74ze/pr74ze-dss-0001/ra95kix2/results/cpp/decorte/skills_only_pjmath_desc_weighted_idf_logpool_adv/scores/test_scores_text.pkl'

def load_targets(path):
    with open(path, "rb") as f:
        d = pickle.load(f)
    # assume compatible file structure: check 'target_labels' or fallback
    if 'target_labels' in d:
        return set(d['target_labels'])
    elif 'target_label' in d:
        return set(d['target_label'])
    elif hasattr(d, 'target_labels'):
        return set(d.target_labels)
    else:
        raise ValueError(f"No target_labels key in {path}")

text_targets = load_targets(text_path)
hybrid_targets = load_targets(hybrid_path)
skill_targets = load_targets(skill_path)

print("Number of unique labels:")
print("Text:", len(text_targets))
print("Hybrid:", len(hybrid_targets))
print("Skill:", len(skill_targets))

if text_targets == hybrid_targets == skill_targets:
    print("All have identical target label pools.")
else:
    print("Target label pools differ!")
    print("Only in text:", text_targets - hybrid_targets - skill_targets)
    print("Only in hybrid:", hybrid_targets - text_targets - skill_targets)
    print("Only in skill:", skill_targets - text_targets - hybrid_targets)
    print("Intersection size:", len(text_targets & hybrid_targets & skill_targets))
    print("Difference (text vs hybrid):", len(text_targets.symmetric_difference(hybrid_targets)))
    print("Difference (hybrid vs skill):", len(hybrid_targets.symmetric_difference(skill_targets)))
    print("Difference (text vs skill):", len(text_targets.symmetric_difference(skill_targets)))
```
