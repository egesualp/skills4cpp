```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'/dss/dsshome1/02/ra95kix2/thesis/skills4cpp/data/processed/augmentation/augmented_decorte_occupations_with_desc_2.csv')
```

```python
# Tokenization statistics for skill_brief column
from transformers import AutoTokenizer

# Load tokenizer (using the same model as in the similarity scorer)
tokenizer = AutoTokenizer.from_pretrained("BAAI/bge-large-en-v1.5")

# Tokenize skill_brief column
df['skill_brief_tokens'] = df['skill_brief'].apply(lambda x: len(tokenizer.encode(str(x), add_special_tokens=True)) if pd.notna(x) else 0)

# Calculate statistics
token_stats = df['skill_brief_tokens'].describe(percentiles=[0.25, 0.5, 0.75, 0.9, 0.95, 0.99])
print("Token count statistics for skill_brief column:")
print(token_stats)
print(f"\nMedian: {df['skill_brief_tokens'].median()}")
print(f"Mean: {df['skill_brief_tokens'].mean():.2f}")
```
